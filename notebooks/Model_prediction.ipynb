```python
import pandas as pd
import numpy as np
import joblib
```

```python
best_model = joblib.load(
    r"D:\Vehicle_Mileage_Project\models\best_model.pkl"
)

label_encoders = joblib.load(
    r"D:\Vehicle_Mileage_Project\models\label_encoders.pkl"
)

feature_names = joblib.load(
    r"D:\Vehicle_Mileage_Project\models\feature_names.pkl"
)
```

```python
new_vehicle = pd.DataFrame({
    "Vehicle_Age": [5],
    "Engine_CC": [1200],
    "Horsepower": [95],
    "Vehicle_Weight": [1100],
    "Transmission": ["Manual"],
    "Ethanol_Blend": [20],
    "Fuel_Price": [104],
    "Trip_Distance_km": [150],
    "Average_Speed": [65],
    "Engine_RPM": [2400],
    "Tire_Pressure_PSI": [32],
    "Maintenance_Status": ["Good"],
    "Road_Type": ["City"],
    "Traffic_Level": ["Medium"],
    "Temperature": [30],
    "Humidity": [60],
    "Weather": ["Clear"],
    "AC_Usage": ["Yes"],
    "Driving_Style": ["Normal"],
    "Passengers": [3]
})
```

```python
print(label_encoders.keys())
```

```python
categorical_columns = [
    "Transmission",
    "Maintenance_Status",
    "Road_Type",
    "Traffic_Level",
    "Weather",
    "AC_Usage",
    "Driving_Style"
]

for col in categorical_columns:
    new_vehicle[col] = label_encoders[col].transform(new_vehicle[col])
```

```python
new_vehicle = new_vehicle[feature_names]
```

```python
prediction = best_model.predict(new_vehicle)

print(f"Predicted Mileage: {prediction[0]:.2f} km/L")
```
